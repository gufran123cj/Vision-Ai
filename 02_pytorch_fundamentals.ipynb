## Görev 1 — Tensor 101
- Rastgele tensör oluşturma, dilimleme, yeniden şekillendirme
- Matris çarpımı (`matmul`)
- Autograd ile türev takibi


```python
import torch

# Tensor oluşturma
x = torch.randn(3, 2, dtype=torch.float32)
w = torch.randn(2, 1, requires_grad=True)
b = torch.randn(1, requires_grad=True)

print('x shape:', x.shape)
print('w dtype:', w.dtype)
print('b requires_grad:', b.requires_grad)

# Basit lineer işlem
z = x @ w + b
print('z:', z)

# Autograd örneği
loss = z.sum()
loss.backward()

print('w.grad:', w.grad)
print('b.grad:', b.grad)

# requires_grad kapatma
with torch.no_grad():
    z2 = x @ w + b
    print('Grad takibi yok:', z2)
```

## Tensor işlemleri — view/reshape


```python
t = torch.arange(12)
print('Orijinal:', t)
print('view(3,4):\n', t.view(3,4))
print('reshape(2,6):\n', t.reshape(2,6))
```
